# 传统方式
## ReAct 模式

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.tools import Tool
from langchain.agents import Agent, AgentType, initialize_agent, create_tool_calling_agent, AgentExecutor, \
    create_react_agent
from langchain_community.tools import TavilySearchResults
from langchain_openai import ChatOpenAI
import dotenv
import os

dotenv.load_dotenv()

llm = ChatOpenAI(model=os.getenv("OPENAI_MODEL"), temperature=0)

# 创建工具
search = TavilySearchResults(max_results = 3)
# tools = [search]

searchTool = Tool(
    func=search.run,
    name="Search",
    description="用于查询互联网上的信息"
)
tools = [searchTool]

# 绑定工具
llm_with_tools = llm.bind_tools(tools)

# 获取 Agent
agent = AgentType.ZERO_SHOT_REACT_DESCRIPTION

# 构建 AgentExecutor
agentExecutor = initialize_agent(
    tools = tools,
    llm = llm_with_tools,
    agent = agent,
    verbose = True
)
# 调用模型
response = agentExecutor.invoke("查询北京今天的天气情况")

print(response)
```
