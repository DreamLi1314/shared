# 文档加载器

```python
from langchain_community.document_loaders import JSONLoader

json_loader = JSONLoader(
    file_path="assets/json.json",
    jq_schema=".", # 表示加载所有的字段
    text_content=False
)

docs = json_loader.load()

print(docs)
```

```python
json_loader = JSONLoader(
    file_path="assets/json.json",
    jq_schema=".arr[].age", # 表示加载所有的字段
    text_content=False
)

docs = json_loader.load()

print(docs)
```

## 失败

```python
json_loader = JSONLoader(
    file_path="assets/json.json",
    jq_schema=".arr[]",
    # text_content=False,
    content_key='.name + "\n\n" + .age',
    is_content_key_jq_parsable=True
)

docs = json_loader.load()

print(docs)
```
