# @ tool

```python
from langchain_core.tools import tool, StructuredTool
from pydantic import BaseModel, Field


@tool
def add_number(a:int, b:int) -> int:
    """计算连个数的和"""
    return a+b

print(f"name= {add_number.name}")
print(f"args= {add_number.args}")
print(f"description= {add_number.description}")
print(f"return_direct= {add_number.return_direct}")
```

## 修改工具元数据

```python
from langchain_core.tools import tool

@tool(name_or_callable="add_two_number", description="计算两个数字的和")
def add_number(a:int, b:int) -> int:
    """计算连个数的和"""
    return a+b

print(f"name= {add_number.name}")
print(f"args= {add_number.args}")
print(f"description= {add_number.description}")
print(f"return_direct= {add_number.return_direct}")
```

## 调用工具

```python
add_number.invoke({"a": 10, "b": 30})
```

## 修改 args

```python
from langchain_core.tools import tool

class FieldInfo(BaseModel):
    a:int = Field(description="第一个参数")
    b:int = Field(description="第二个参数")

@tool(name_or_callable="add_two_number", description="计算两个数字的和", args_schema=FieldInfo)
def add_number(a:int, b:int) -> int:
    """计算连个数的和"""
    return a+b

print(f"name= {add_number.name}")
print(f"args= {add_number.args}")
print(f"description= {add_number.description}")
print(f"return_direct= {add_number.return_direct}")
```

# StructuredTool

```python
class FieldInfo(BaseModel):
    query:str = Field(description="第一个参数")

def search_google(query: str):
    return f"这是查询{query}的结果"

search01 = StructuredTool.from_function(
    func=search_google,
    name="searchGoogle",
    description="搜索谷歌, 并将结果返回",
    return_direct=True,
    args_schema=FieldInfo
)

print(f"name= {search01.name}")
print(f"args= {search01.args}")
print(f"description= {search01.description}")
print(f"return_direct= {search01.return_direct}")
```

```python
search01.invoke({"query": "aaa"})
```
