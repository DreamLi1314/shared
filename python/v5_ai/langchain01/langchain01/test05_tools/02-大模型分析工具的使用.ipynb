 # 大模型分析

```python
from langchain_community.tools import MoveFileTool
from langchain_core.messages import HumanMessage
from langchain_openai import ChatOpenAI
import dotenv
import os

dotenv.load_dotenv()

llm = ChatOpenAI(model=os.getenv("OPENAI_MODEL"), temperature=0)

# 获取工具列表
tools = [MoveFileTool()]

msgs = [HumanMessage(content="将当前目录下的文件moveTes.txt移动到/Users/dreamli/Workspace/shared/python/v5_ai/langchain01/langchain01")]

#  为模型绑定工具 
llm_with_tools = llm.bind_tools(tools)
response = llm_with_tools.invoke(input = msgs)

print(response)
```

# 调用工具

```python
if hasattr(response, 'tool_calls') and response.tool_calls:
    print("模型决定调用工具！")
    for tool_call in response.tool_calls:
        tool_name = tool_call["name"]
        tool_args = tool_call["args"]
        print(f"工具名: {tool_name}, 参数: {tool_args}")

        # 找到被调用的工具并执行
        tool_to_use = {tool.name: tool for tool in tools}.get(tool_name)
        print(f"tool_to_use: {tool_to_use}")
        if tool_to_use:
            try:
                # 执行工具
                tool_result = tool_to_use.invoke(tool_args)
                print(f"工具执行结果: {tool_result}")
            except Exception as e:
                print(f"工具调用出错: {e}")
        else:
            print(f"未找到名为 {tool_name} 的工具")
else:
    print("模型未调用工具，直接回复：", response.content)
```
