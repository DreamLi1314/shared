#  JsonOutputParser 

```python
import os

from dotenv import load_dotenv
from langchain_core.output_parsers import JsonOutputParser, XMLOutputParser
from langchain_core.prompts import ChatPromptTemplate, PromptTemplate
from langchain_openai import ChatOpenAI

load_dotenv()

llm  = ChatOpenAI(model = os.getenv("OPENAI_MODEL"))

template = ChatPromptTemplate.from_messages([
    ("system", "你是一个专业的{role}"),
    ("human", "{que},问题用q表示，答案用a表示，返回一个JSON格式")
])

prompt = template.invoke({
    "role": "翻译官",
    "que": "打架用英语怎么说?"
})

response = llm.invoke(prompt)
print(type(response))
print(type(response.content))
print(response.content)

print("======")

parser = JsonOutputParser()
jsonResult = parser.invoke(response)
print(type(jsonResult))
print(jsonResult)
```

# JsonOutputParser2: Chain

```python
import os

from dotenv import load_dotenv
from langchain_core.output_parsers import JsonOutputParser
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI

load_dotenv()

llm  = ChatOpenAI(model = os.getenv("OPENAI_MODEL"))

template = ChatPromptTemplate.from_messages([
    ("system", "你是一个专业的{role}"),
    ("human", "{que},问题用q表示，答案用a表示，返回一个JSON格式")
])

parser = JsonOutputParser()

# prompt = template.invoke({
#     "role": "翻译官",
#     "que": "打架用英语怎么说?"
# })
# response = llm.invoke(prompt)
# jsonResult = parser.invoke(response)

chain = template | llm | parser
jsonResult = chain.invoke({
    "role": "翻译官",
    "que": "打架用英语怎么说?"
})

print(type(jsonResult))
print(jsonResult)
```

# XML2

```python
xmlParser  =  XMLOutputParser()
actor_query = "周星驰的电影记录"

template =  PromptTemplate.from_template("生成{actor_query},格式为:{format}")
template = template.partial(format=xmlParser.get_format_instructions())

response = llm.invoke(template.invoke({"actor_query": actor_query}))
print(type(response))
print(response.content)
```
