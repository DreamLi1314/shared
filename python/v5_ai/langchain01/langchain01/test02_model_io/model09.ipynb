# 模板示例

```python
import os

from langchain_core.prompts import FewShotPromptTemplate, PromptTemplate, FewShotChatMessagePromptTemplate, \
    ChatPromptTemplate
from langchain_openai import ChatOpenAI
import dotenv

dotenv.load_dotenv()

'''
提示词模板
'''

llm = ChatOpenAI(model=os.getenv("OPENAI_MODEL"),
                 streaming=False)

res = llm.invoke("2 🦜 9是多少?")
print(res.content)
```

```python
prompt = PromptTemplate.from_template("input:{input},output:{output}")

examples = [
    {"input": "北京天气怎么样", "output": "北京市"},
    {"input": "南京下雨吗", "output": "南京市"},
    {"input": "武汉热吗", "output": "武汉市"}
]

fewPrompt =  FewShotPromptTemplate(
    example_prompt=prompt,
    examples=examples,
    suffix="input:{input}\noutput:",
    input_variables=["input"]
)

prom=fewPrompt.invoke({
    "input": "天津会下雨吗?"
})

res = llm.invoke(prom)
print(res)
```

```python
prompt = ChatPromptTemplate.from_messages([
    ("human", "{input}是多少?"),
    ("ai", "{output}")
])

examples = [
    {"input": "2 🦜 9", "output": "11"},
    {"input": "3 🦜 7", "output": "10"},
    {"input": "1 🦜 0", "output": "1"}
]

fewChatPrompt = FewShotChatMessagePromptTemplate(
    example_prompt=prompt,
    examples=examples,
)

finalPrompt = ChatPromptTemplate.from_messages([
    ("system", "你是一个数学专家"),
    fewChatPrompt,
    ("human", "{input}"),
])

res = llm.invoke(finalPrompt.invoke({
    "input": "9 🦜 9"
}))
print(res.content)
```
