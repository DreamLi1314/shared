# 提示词模板的使用

## 1

```python
from langchain_core.prompts import PromptTemplate

promptTemplate = PromptTemplate(
    template = "你是一个{role}, 你的名字是{name}",
    input_variables = ["role", "name"]
)
print(promptTemplate)

prompt = promptTemplate.format(role="人工智能专家", name = "小智")
print(prompt)
```

## 2 推荐方式

```python
from langchain_core.prompts import PromptTemplate

promptTemplate = PromptTemplate.from_template("你是一个{role}, 你的名字是{name}")
print(promptTemplate)

prompt = promptTemplate.format(role="人工智能专家", name = "小智")
print(prompt)
```

## 3

```python
from langchain_core.prompts import PromptTemplate

promptTemplate = PromptTemplate.from_template(
    template = "你是一个{role}, 你的名字是{name}",
    partial_variables = { "role": "人工智能专家"}
)
print(promptTemplate)

prompt = promptTemplate.format(name = "小智")
print(prompt)
```

## 4 推荐

```python
from langchain_core.prompts import PromptTemplate

promptTemplate = PromptTemplate.from_template(
    template = "你是一个{role}, 你的名字是{name}",
    partial_variables = { "role": "人工智能专家"}
)
print(promptTemplate)

prompt = promptTemplate.invoke(input = {"name": "小智"})
print(type(prompt))
print(prompt)
print(prompt.text)
```
